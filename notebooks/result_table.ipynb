```python
from stt_toolkit import ResultCache

model_names = [
    "openai/whisper-large-v3-turbo",
    "whispercpp/large-v3-turbo",
    "bofenghuang/whisper-large-v3-distil-it-v0.2",
    "whispercpp/whisper-large-v3-distil-it-v0.2",
    "whispercpp/whisper-large-v3-distil-it-v0.2-q5_0",
]
tasks = ["lopozz/accenti_italiani[train]"]

cache = ResultCache("../results")
results = cache.load_results(
    models=model_names,
    tasks=tasks,
)

results.to_dataframe()
```
